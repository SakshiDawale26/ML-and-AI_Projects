```python
from playwright.async_api import async_playwright
import asyncio

async def fill_form():
    async with async_playwright() as p:
        # Launch browser in headless mode
        browser = await p.chromium.launch(headless=True)
        page = await browser.new_page()
        await page.goto("https://httpbin.org/forms/post")

        # Fill text inputs
        await page.fill("input[name='custname']", "John Doe")
        await page.fill("input[name='custtel']", "555-123-4567")
        await page.fill("input[name='custemail']", "john.doe@example.com")

        # Wait until the dropdown exists and set value using JS
        await page.wait_for_selector("select[name='size']", timeout=10000)
        await page.evaluate("""() => {
            const el = document.querySelector("select[name='size']");
            if (el) el.value = "medium";
        }""")

        # Check the topping checkbox
        await page.wait_for_selector("input[name='topping'][value='cheese']", timeout=5000)
        await page.check("input[name='topping'][value='cheese']")

        # Fill remaining fields
        await page.fill("input[name='delivery']", "now")
        await page.fill("textarea[name='comments']", "This is a test form submission")

        # Submit the form
        await page.click("input[type='submit']")

        # Wait a bit and print response page content
        await page.wait_for_timeout(500)
        content = await page.content()
        print(content)

        # Close browser
        await browser.close()

# Run the async function
asyncio.run(fill_form())
```

```python
from playwright.async_api import async_playwright
import asyncio

async def run_basic_search(query="OpenAI"):
    visited_urls = []

    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=True)
        context = await browser.new_context(
            user_agent="Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                       "AppleWebKit/537.36 (KHTML, like Gecko) "
                       "Chrome/116.0.0.0 Safari/537.36",
            viewport={"width": 1280, "height": 800}
        )

        page = await context.new_page()

        # Go to DuckDuckGo HTML version
        await page.goto(f"https://duckduckgo.com/html/?q={query}")

        # Small wait to ensure the page loads
        await page.wait_for_timeout(1000)

        # Grab all <a> elements inside results
        links = await page.query_selector_all("a")
        for link in links:
            url = await link.get_attribute("href")
            if url and url.startswith("http"):
                visited_urls.append(url)

        await browser.close()

    return visited_urls

# Run async function
history_urls = asyncio.run(run_basic_search("OpenAI"))
print(history_urls)
```

```python
from playwright.async_api import async_playwright
import asyncio

class SearchHistory:
    def __init__(self):
        self._history = []

    def add_urls(self, query, urls):
        for url in urls:
            self._history.append({"query": query, "url": url})

    def urls(self):
        # Returns list of URLs only
        return [entry["url"] for entry in self._history]

    def full_history(self):
        # Returns full history with queries
        return self._history

async def run_search_queries(queries):
    history = SearchHistory()

    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=True)
        context = await browser.new_context(
            user_agent="Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                       "AppleWebKit/537.36 (KHTML, like Gecko) "
                       "Chrome/116.0.0.0 Safari/537.36",
            viewport={"width": 1280, "height": 800}
        )

        page = await context.new_page()

        for query in queries:
            # Go to DuckDuckGo HTML version
            await page.goto(f"https://duckduckgo.com/html/?q={query}")

            # Short wait for page load
            await page.wait_for_timeout(1000)

            # Grab all links
            links = await page.query_selector_all("a")
            urls = []
            for link in links:
                url = await link.get_attribute("href")
                if url and url.startswith("http"):
                    urls.append(url)

            # Add URLs to history
            history.add_urls(query, urls)

        await browser.close()

    return history

# Example usage
queries = ["OpenAI", "ChatGPT", "Playwright Python"]

search_history = asyncio.run(run_search_queries(queries))

# Get all URLs only
print("All URLs:", search_history.urls())

# Get full history with queries
print("Full history:", search_history.full_history())
```
