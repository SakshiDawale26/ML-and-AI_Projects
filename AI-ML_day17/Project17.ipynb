* **Open and Deployable:** As a fully open-source solution, Crawl4AI requires no API keys and is ready for immediate integration into cloud or on-premise environments using Docker.

```python
# Install dependencies if not already installed
!pip install crawl4ai nest_asyncio playwright
!playwright install


import asyncio
import nest_asyncio
from crawl4ai import AsyncWebCrawler, CrawlerRunConfig
from crawl4ai.deep_crawling import BFSDeepCrawlStrategy
from crawl4ai.content_scraping_strategy import LXMLWebScrapingStrategy

# Apply nest_asyncio for notebooks/Colab
nest_asyncio.apply()

async def main():
    # Configure crawler
    config = CrawlerRunConfig(
        deep_crawl_strategy=BFSDeepCrawlStrategy(
            max_depth=1,          # Change this for deeper crawl
            include_external=False # Only stay within same domain
        ),
        scraping_strategy=LXMLWebScrapingStrategy(),
        verbose=True
    )

    async with AsyncWebCrawler() as crawler:
        results = await crawler.arun("https://www.wikipedia.org", config=config)
        print(f"\nCrawled {len(results)} pages in total\n")

        # Display first few results
        for idx, result in enumerate(results[:5]):  # Show first 5 results
            print(f"--- Result {idx+1} ---")
            print(f"URL: {result.url}")
            print(f"Depth: {result.metadata.get('depth', 0)}")
            print("MARKDOWN CONTENT:\n", result.markdown[:500], "...")  # first 500 chars
            print("\n")

# Run the async main function
await main()
```

#Initial Observation (with max_depth=2)

When the crawler was initially set to max_depth=2, the BFS algorithm attempted to visit all pages at Depth 1 before exploring Depth 2 pages.

On https://www.wikipedia.org, this generated a queue of over 300 pages at Depth 1.

As a result, the crawl ran indefinitely and did not finish within a practical time frame.

Although technically finite, this depth caused a very long runtime, making it impractical for generating results for the assignment.

# Modification for Practical Results (max_depth=1)

To make the crawl manageable and demonstrable, the max_depth parameter was reduced from 2 to 1.

This modification instructed the crawler to:

Visit the main portal page (Depth 0).

Discover all links on that page.

Visit a limited number of those discovered pages (Depth 1) and then stop.

Outcome:

Successfully crawled 343 pages.

Crawl finished in approximately 9 minutes.

Allowed demonstration of BFS crawling behavior without overwhelming the system.

# Summary

BFS crawler processes all pages at one depth before moving deeper.

Setting a high max_depth can lead to excessive crawling time if the site has many links.

Adjusting max_depth is necessary to balance depth coverage and runtime.

The final configuration demonstrates:

How BFS crawls pages in a breadth-first manner.

How metadata (URL, depth) can be captured for each crawled page.

Practical scraping results within reasonable time limits.
