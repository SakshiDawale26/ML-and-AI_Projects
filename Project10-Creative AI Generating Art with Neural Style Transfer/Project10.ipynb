```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import os

# Step 1: Load pretrained generator
generator = tf.keras.models.load_model("generator_64.h5")
print("Generator loaded!")

# Step 2: Define a dummy gender vector (in practice, this comes from CelebA attributes / pretrained embeddings)
# Here we simulate it as a random direction in the latent space
np.random.seed(42)
gender_vector = np.random.normal(0, 1, (NOISE_DIM,))
gender_vector = gender_vector / np.linalg.norm(gender_vector)  # normalize

print("Gender vector loaded (simulated). Shape:", gender_vector.shape)

# Step 3: Pick one base random latent code (z)
z_base = np.random.normal(0, 1, (1, NOISE_DIM))

# Step 4: Generate 10 variations along gender vector
alphas = np.linspace(-2, 2, 10)   # interpolation strength
generated_images = []

for alpha in alphas:
    z_new = z_base + alpha * gender_vector  # move along gender direction
    img = generator.predict(z_new, verbose=0)
    img = (img[0] + 1.0) / 2.0  # rescale from [-1,1] → [0,1]
    generated_images.append(img)

# Step 5: Plot results
plt.figure(figsize=(20, 4))
for i, img in enumerate(generated_images):
    plt.subplot(1, 10, i+1)
    plt.imshow(img)
    plt.axis("off")
    plt.title(f"{alphas[i]:.1f}")
plt.suptitle("Male → Female Interpolation", fontsize=18)
plt.show()
```
